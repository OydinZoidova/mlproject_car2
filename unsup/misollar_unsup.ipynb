```python
from sklearn.cluster import KMeans
import numpy as np
# clientning oylik maosji, dukonga necha marta tashrif buyurishi
mijozlar_malumoti=np.array([
    [200,3],[220,4],[50,1],
    [60,2],[800,10],[850,12]
])
klaster_soni=3
model=KMeans(n_clusters=klaster_soni, random_state=42)
model.fit(mijozlar_malumoti)
guruhlar=model.labels_
markazlar=model.cluster_centers_
print("Mijoz tushgan har bir guruh: ", guruhlar)
```

```python
from sklearn.decomposition import PCA # pca- klasterlash uchun xususiyatlar ichidan 2,3 ta asosiylarini ajratib olish
xususiyatlar=np.array([
    [90,95,8,9],[85,90,7,8],
    [50,40,2,3], [35,45,2,3]
    
])
pca=PCA(n_components=2)
siqilgan_malumot=pca.fit_transform(xususiyatlar)
print("2 o'lchamli siqilgan ma'lumot: ", siqilgan_malumot)
print("Yangi soddalshgan qiymat: ", siqilgan_malumot)
```
